## Syntax 

Python syntax uses indentations 

if condition: 
    logic
else: 
    logic 


```python
exampleInput = int(input("pick a number. any number: ")) 

if exampleInput > 5: 
    print("that number is at least greater than 5")
else: 
    print("Smol boy number") 
```

**Bonus**: They can be chaaaained

```python
if exampleInput > 5: 
    print("that number is at least greater than 5")
    if exampleInput > 10:
        print("that number is at least greater than 10")
    else: 
        print("less than 10 but greater than 5") 
    if exampleInput > 15:
        print("that number is at least greater than 15")
    else: 
        print("less than 15 but greater than 10")
else: 
    print("Smol boy number") 
```

Only the first condition met is considered if not chained properly. ex: 

```python
if exampleInput > 5: 
    print("that number is at least greater than 5")
elif exampleInput > 10:
    print("that number is at least greater than 10")
elif exampleInput > 15:
    print("that number is at least greater than 15")
else: 
    print("Smol boy number") 
```

## Type conversions 

There are 2 basic types
- strings - alphanumeric
- integers - whole numbers
- floats - decimal numbers

Type conversions can be done to change the type of a variable 

```python
variableExample = "78" 
variableIntExample = int(variableExample) 
variableFloatExample = float(variableExample) 
print(type(variableExample))
print(type(variableIntExample))
print(type(variableFloatExample))
```
