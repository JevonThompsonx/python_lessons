## Conversions

The 3 basic types are 

- integers - int
- numbers with decimal points - float
- alphanumerical characters - strings

Conversion between these types can be done by calling the type and wrapping a variable/value in that invocation 

eg: 


```python
ranString = "22" 
ranNum = int(ranString) 
print(type(ranString), ranString) 
print(type(ranNum), ranNum) 
```
